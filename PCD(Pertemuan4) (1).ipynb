```python
import cv2

#Open image
gambar = cv2.imread("/content/bawang.jpg")

#manipulasi citra (arg: gambar_rotate)
gambar_rotate = cv2.rotate(gambar, cv2.ROTATE_90_CLOCKWISE)

#manipulasi citra (arg: img_flip, 1=horizontal)
gambar_flip = cv2.flip(gambar, 1)

h, w = gambar.shape[:2]
center = (w/2, h/2)
angle = 35
scale = 1
matrix = cv2.getRotationMatrix2D(center, angle, scale)

#gambar rotate
gambar_rotate = cv2.warpAffine(gambar, matrix, (w, h))

gambar_resize = cv2.resize(gambar, (200, 200))

gambar_rotate = cv2.warpAffine(gambar, matrix, (w, h))

#crop gambar (Y1,Y2 dan X1,X2)
gambar_crop = gambar[100:200, 100:200]


fig, ax = plt.subplots(2,2, figsize=(10,10))
ax[0, 0].imshow(gambar_flip)
ax[0, 0].set_title("Flip")
ax[0, 1].imshow(gambar_rotate)
ax[0, 1].set_title("Rotate")
ax[1, 0].imshow(gambar_resize)
ax[1, 0].set_title("Resize")
ax[1, 1].imshow(gambar_crop)
ax[1, 1].set_title("Crop")

plt.show()
```

```python
import matplotlib.image as mpimg

#OPen data menggunakan opencv
gambar_cv = cv2.imread("/content/bawang.jpg")

#OPen data menggunakan mpimg
gambar_plt  = mpimg.imread("/content/bawang.jpg")

fig, ax = plt.subplots(1, 2, figsize=(10,10))

ax[0].imshow(gambar_cv)
ax[0].set_title("Opencv")

ax[1].imshow(gambar_plt)
ax[1].set_title("Matplotlib")
```

```python
#gambar cv(BGR)
gambarcv = gambar_cv.copy()

#gambar cv(RGB)
gambarcv_rgb = cv2.cvtColor(gambarcv, cv2.COLOR_BGR2RGB)

#gambar HSV (hue, saturatuin, value)
gambarcv_hsv = cv2.cvtColor(gambarcv, cv2.COLOR_BGR2HSV)

#gambar gray (grayscale)
gambar_cv_gray = cv2.cvtColor(gambarcv, cv2.COLOR_BGR2GRAY)

#gambar binary (biner)
_, binary = cv2.threshold(gambar_cv_gray, 128, 255, cv2.THRESH_BINARY)

fig, ax = plt.subplots(2, 2, figsize=(16,16))
ax[0,0].imshow(gambarcv_rgb)
ax[0,0].set_title("opencv_rgb")

ax[0,1].imshow(gambarcv_hsv)
ax[0,1].set_title("Opencv_HSV")

ax[1,0].imshow(gambar_cv_gray, cmap="gray")
ax[1,0].set_title("Opencv_gray")

ax[1,1].imshow(binary, cmap="gray")
ax[1,1].set_title("Opencv_binary")

plt.tight_layout()
plt.show()
```

```python
#Kualitas gambar dipengaruhi oleh"
#RESOLUSI

gambar_small = cv2.resize(gambarcv_rgb, (50, 50))
gambar_big = cv2.resize(gambarcv_rgb, (500, 500))

print("Resolusi Asli", gambarcv_rgb.shape)
print("Resolusi kecil", gambar_small.shape)
print("Resolusi besar", gambar_big.shape)

fig, ax = plt.subplots(1, 3, figsize=(16, 8))
ax[0].imshow(gambarcv_rgb)
ax[0].set_title("Resolusi Asli")
ax[1].imshow(gambar_small)
ax[1].set_title("Resolusi Kecil")
ax[2].imshow(gambar_big)
ax[2].set_title("Resolusi Besar")

plt.tight_layout()
plt.show()
```

```python
#NOISE
noise = np.random.normal(0, 25, gambarcv_rgb.shape).astype("uint8")
noise_gambar = cv2.add(gambarcv_rgb, noise)

blur_gambar = cv2.GaussianBlur(gambarcv_rgb, (5, 5), 0)

fig, ax = plt.subplots(1, 3, figsize=(10, 10))
ax[0].imshow(gambarcv_rgb)
ax[0].set_title("Gambar Asli")
ax[1].imshow(noise_gambar)
ax[1].set_title("Gambar dengan Noise")
ax[2].imshow(blur_gambar)
ax[2].set_title("Gambar dengan Blur")

plt.tight_layout()
plt.show()
```

```python
import os

cv2.imwrite("bawang_90.jpg", cv2.cvtColor(gambarcv_rgb, cv2.COLOR_RGB2BGR),[cv2.IMWRITE_JPEG_QUALITY, 90])
cv2.imwrite("bawang_60.jpg", cv2.cvtColor(gambarcv_rgb, cv2.COLOR_RGB2BGR),[cv2.IMWRITE_JPEG_QUALITY, 60])
cv2.imwrite("bawang_30.jpg", cv2.cvtColor(gambarcv_rgb, cv2.COLOR_RGB2BGR),[cv2.IMWRITE_JPEG_QUALITY, 30])

gambar90 = cv2.cvtColor(cv2.imread("bawang_90.jpg"), cv2.COLOR_BGR2RGB)
gambar60 = cv2.cvtColor(cv2.imread("bawang_60.jpg"), cv2.COLOR_BGR2RGB)
gambar30 = cv2.cvtColor(cv2.imread("bawang_30.jpg"), cv2.COLOR_BGR2RGB)

fig, ax = plt.subplots(1, 3, figsize=(10, 10))
ax[0].imshow(gambar90)
ax[0].set_title("Kualitas 90%")
ax[1].imshow(gambar60)
ax[1].set_title("Kualitas 60%")
ax[2].imshow(gambar30)
ax[2].set_title("Kualitas 30%")

print(os.path.getsize("bawang_90.jpg"))
print(os.path.getsize("bawang_60.jpg"))
print(os.path.getsize("bawang_30.jpg"))

plt.tight_layout()
plt.show()
```

```python
#menghitung tingkat error dari citra

from skimage.metrics import mean_squared_error

mse90 = mean_squared_error(gambarcv_rgb, gambar90)
mse60 = mean_squared_error(gambarcv_rgb, gambar60)
mse30 = mean_squared_error(gambarcv_rgb, gambar30)

print("MSE Gambar 90% : ", mse90)
print("MSE Gambar 60% : ", mse60)
print("MSE Gambar 30% : ", mse30)

#semakin rendah tingkat MSE maka semakin bagus tnigkat gambar nya
```

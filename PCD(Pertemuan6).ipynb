```python
from scipy import datasets, ndimage
from matplotlib import pyplot as plt

face = datasets.face()
blurred_face = ndimage.gaussian_filter(face, sigma=3)

fig, ax = plt.subplots(1, 2, figsize = (10, 5))

ax[0].imshow(face)
ax[0].set_title('Gambar Asli')
# ax[0].axis('off')
ax[0].set_xticks([])
ax[0].set_yticks([])

ax[1].imshow(blurred_face)
ax[1].set_title('Gambar Blur')
# ax[1].axis('off')
ax[1].set_xticks([])
ax[1].set_yticks([])

plt.show()
```

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

gambar = cv.imread("/content/jambu.jpg")

fig, ax = plt.subplots(1, 3, figsize = (10, 5))
fig.tight_layout()

#to conovole the kernel on an image we can use cv.filter2D

ax[0].imshow(cv.cvtColor(gambar, cv.COLOR_BGR2RGB))
ax[0].set_title('Gambar Asli')

kernel_sharpening = np.array([[-1,-1,-1],
                              [-1,9,-1],
                              [-1,-1,-1]])
kernel_sharpening2 = np.array([[-1,-1,-1],
                               [-1,10,-1],
                               [-1, -1, -1]])

sharpened = cv.filter2D(gambar, -1, kernel_sharpening)
ax[1].imshow(cv.cvtColor(sharpened, cv.COLOR_BGR2RGB))
ax[1].set_title('Gambar Sharpened')

sharpened2 = cv.filter2D(gambar, -1, kernel_sharpening2)
ax[2].imshow(cv.cvtColor(sharpened2, cv.COLOR_BGR2RGB))
ax[2].set_title('Gambar Sharpened2')

plt.show()
#
```

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

gambar = cv.imread("/content/jambu.jpg", cv.IMREAD_GRAYSCALE)

kernel_size = 3 #bisa 5, 7, 9
kernel = np.ones((kernel_size, kernel_size), np.uint8)
min_filter = cv.erode(gambar, kernel, iterations=1)

plt.figure(figsize=(10, 5))

plt.subplot(2, 2, 1)
plt.imshow(gambar, cmap="gray")
plt.title("Gambar Asli")

plt.subplot(2, 2, 2)
plt.imshow(min_filter, cmap="gray")
plt.title("Gambar Min Filter")

plt.subplot(2,2,3)
plt.hist(gambar.ravel(), 256, [0, 256], color="blue")
plt.title("Histogram Gambar Asli")
plt.xlabel("Tingkat keabuan [0-255]")
plt.ylabel("Frekuensi")

plt.subplot(2,2,4)
plt.hist(min_filter.ravel(), 256, [0, 256], color="blue")
plt.title("Histogram Gambar Min Filter")
plt.xlabel("Tingkat keabuan [0-255]")
plt.ylabel("Frekuensi")

plt.tight_layout()
plt.show()

#min filter = mengubah gambar menjadi lebih gelap
```

```python
#max filter juga menggunakan kernel, kernel nya kita gunakan
max_filter = cv.dilate(gambar, kernel)

plt.figure(figsize=(10, 5))

plt.subplot(2, 2, 1)
plt.imshow(gambar, cmap="gray")
plt.title("Gambar Asli")

plt.subplot(2, 2, 2)
plt.imshow(max_filter, cmap="gray")
plt.title("Gambar Max Filter")

plt.subplot(2,2,3)
plt.hist(gambar.ravel(), 256, [0, 256], color="blue")
plt.title("Histogram Gambar Asli")
plt.xlabel("Tingkat keabuan [0-255]")
plt.ylabel("Frekuensi")

plt.subplot(2,2,4)
plt.hist(max_filter.ravel(), 256, [0, 256], color="blue")
plt.title("Histogram Gambar Max Filter")
plt.xlabel("Tingkat keabuan [0-255]")
plt.ylabel("Frekuensi")

plt.tight_layout()
plt.show()
#max filter berguan untuk mencerahkan gambar
```

```python
mean_filter = cv.blur(gambar, (kernel_size, kernel_size))

plt.figure(figsize=(10, 5))

plt.subplot(2, 2, 1)
plt.imshow(gambar, cmap="gray")
plt.title("Gambar Asli")

plt.subplot(2, 2, 2)
plt.imshow(mean_filter, cmap="gray")
plt.title("Gambar Mean Filter")

plt.subplot(2, 2, 3)
plt.hist(gambar.ravel(), 256, [0, 256], color="blue")
plt.title("Histogram Gambar Asli")
plt.xlabel("Tingkat keabuan [0-255]")
plt.ylabel("Frekuensi")

plt.subplot(2, 2, 4)
plt.hist(mean_filter.ravel(), 256, [0, 256], color="blue")
plt.title("Histogram Gambar Mean Filter")
plt.xlabel("Tingkat keabuan [0-255]")
plt.ylabel("Frekuensi")

plt.tight_layout()
plt.show()

#mean filter bertujuan untuk membuat gambar menjadi halus
```

```python
median_filter = cv.medianBlur(gambar, kernel_size)

plt.figure(figsize=(10, 5))

plt.subplot(2, 2, 1)
plt.imshow(gambar, cmap="gray")
plt.title("Gambar Asli")

plt.subplot(2, 2, 2)
plt.imshow(median_filter, cmap="gray")
plt.title("Gambar Median Filter")

plt.subplot(2, 2, 3)
plt.hist(gambar.ravel(), 256, [0, 256], color="blue")
plt.title("Histogram Gambar Asli")
plt.xlabel("Tingkat keabuan [0-255]")
plt.ylabel("Frekuensi")

plt.subplot(2, 2, 4)
plt.hist(median_filter.ravel(), 256, [0, 256], color="blue")
plt.title("Histogram Gambar Median Filter")
plt.xlabel("Tingkat keabuan [0-255]")
plt.ylabel("Frekuensi")

plt.tight_layout()
plt.show()

#median filter bertujuan untuk menghilangkan noise
```

```python
#LPF
#Gasussian

sigma = 1.0
gaussian_filter = cv.GaussianBlur(gambar, (kernel_size, kernel_size), sigma)

plt.figure(figsize=(10, 5))

plt.subplot(2, 2, 1)
plt.imshow(gambar, cmap="gray")
plt.title("Gambar Asli")

plt.subplot(2, 2, 2)
plt.imshow(gaussian_filter, cmap="gray")
plt.title("Gambar Gaussian Filter")

plt.subplot(2, 2, 3)
plt.hist(gambar.ravel(), 256, [0, 256], color="blue")
plt.title("Histogram Gambar Asli")
plt.xlabel("Tingkat keabuan [0-255]")
plt.ylabel("Frekuensi")

plt.subplot(2, 2, 4)
plt.hist(gaussian_filter.ravel(), 256, [0, 256], color="blue")
plt.title("Histogram Gambar Gaussian Filter")
plt.xlabel("Tingkat keabuan [0-255]")
plt.ylabel("Frekuensi")

plt.tight_layout()
plt.show()

#Tapis Lolos Bawah (Low Pass Filter) digunakan untuk menghasluskan gambar dengan mengurangi komponen frekuensi tinggi
```

```python
#Tapis Lolos Atas (High Pass Filter)

sobel = cv.Sobel(gambar, cv.CV_64F, 1, 0, ksize=3)

sobel_x = cv.Sobel(gambar, cv.CV_64F, 1, 0, ksize=3)
sobel_y = cv.Sobel(gambar, cv.CV_64F, 0, 1, ksize=3)

plt.figure(figsize=(10, 5))

plt.subplot(3, 4, 1)
plt.imshow(gambar, cmap="gray")
plt.title("Gambar Asli")

plt.subplot(3, 4, 2)
plt.imshow(sobel, cmap="gray")
plt.title("Gambar Sobel")

plt.subplot(3, 4, 3)
plt.imshow(sobel_x, cmap="gray")
plt.title("Gambar Sobel X")

plt.subplot(3, 4, 4)
plt.imshow(sobel_y, cmap="gray")
plt.title("Gambar Sobel Y")

plt.subplot(3, 4, 5)
plt.hist(gambar.ravel(), 256, [0, 256], color="blue")
plt.title("Histogram Gambar Asli")
plt.xlabel("Tingkat keabuan [0-255]")
plt.ylabel("Frekuensi")

plt.subplot(3, 4, 6)
plt.hist(sobel.ravel(), 256, [0, 256], color="blue")
plt.title("Histogram Gambar Sobel")
plt.xlabel("Tingkat keabuan [0-255]")
plt.ylabel("Frekuensi")

plt.subplot(3, 4, 7)
plt.hist(sobel_x.ravel(), 256, [0, 256], color="blue")
plt.title("Histogram Gambar Sobel X")
plt.xlabel("Tingkat keabuan [0-255]")
plt.ylabel("Frekuensi")

plt.subplot(3, 4, 8)
plt.hist(sobel_y.ravel(), 256, [0, 256], color="blue")
plt.title("Histogram Gambar Sobel Y")
plt.xlabel("Tingkat keabuan [0-255]")
plt.ylabel("Frekuensi")

print(sobel_x)
print(sobel_y)

plt.tight_layout()
plt.show()

#sobel digunakan untuk mendeteksi tepi pada gambar
```

```python
laplacian = cv.Laplacian(gambar, cv.CV_64F)

plt.figure(figsize=(10, 5))

plt.subplot(2, 2, 1)
plt.imshow(gambar, cmap="gray")
plt.title("Gambar Asli")

plt.subplot(2, 2, 2)
plt.imshow(laplacian, cmap="gray")
plt.title("Gambar Laplacian")

plt.subplot(2, 2, 3)
plt.hist(gambar.ravel(), 256, [0, 256], color="blue")
plt.title("Histogram Gambar Asli")
plt.xlabel("Tingkat keabuan [0-255]")
plt.ylabel("Frekuensi")

plt.subplot(2, 2, 4)
plt.hist(laplacian.ravel(), 256, [0, 256], color="blue")
plt.title("Histogram Gambar Laplacian")
plt.xlabel("Tingkat keabuan [0-255]")
plt.ylabel("Frekuensi")

print(laplacian)

plt.tight_layout()
plt.show()

#laplacian berguna untuk mendeteksi garis tepi di semua arah (x dan y)
```

```python
sobel_combine = cv.magnitude(sobel_x, sobel_y)
sobel_uint8 = np.uint8(np.absolute(sobel_combine))
laplacian_uint8 = np.uint8(np.absolute(laplacian))

plt.figure(figsize=(10, 5))

plt.subplot(2, 2, 1)
plt.imshow(sobel_uint8, cmap="gray")
plt.title("Gambar Sobel combine")

plt.subplot(2, 2, 2)
plt.imshow(laplacian_uint8, cmap="gray")
plt.title("Gambar Laplacian Uint8")

plt.subplot(2, 2, 3)
plt.hist(sobel_combine.ravel(), 256, [0, 256], color="blue")
plt.title("Histogram Gambar Sobel Combine")
plt.xlabel("Tingkat keabuan [0-255]")
plt.ylabel("Frekuensi")

plt.subplot(2, 2, 4)
plt.hist(laplacian_uint8.ravel(), 256, [0, 256], color="blue")
plt.title("Histogram Gambar Laplacian Uint8")
plt.xlabel("Tingkat keabuan [0-255]")
plt.ylabel("Frekuensi")

print(sobel_uint8)
print(laplacian_uint8)

plt.tight_layout()
plt.show()
```
